# 📚 Matimo Notebooks - Complete Guide

> **Matimo** gives AI agents 139+ production-ready tools with a built-in Policy Engine, risk classification, and Human-in-the-Loop control. These notebooks need `matimo` 0.2.0 or newer.

Choose a notebook to get started:

## 1️⃣ Quickstart (No API Key)

**[01_quickstart.ipynb](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/01_quickstart.ipynb)** [![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/01_quickstart.ipynb)

Start here if you're new to Matimo. Zero API keys needed.

**What you'll learn:**
- Load Matimo's core tools (provider packages add 139+ more)
- Run a command tool and watch it ask for approval first (secure by default in 0.2.0)
- See the Policy Engine reject unsafe tool definitions, and why
- (Optional) Connect a LangChain agent to Gemini/OpenAI

⏱️ **Time:** 10 min (Part 1 only) or 20 min (with LangChain agent)

## 2️⃣ Policy Engine Deep Dive

**[02_policy_engine.ipynb](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/02_policy_engine.ipynb)** [![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/02_policy_engine.ipynb)

Understand how Matimo governs tools at two moments: when a tool loads, and every time one runs. Zero API keys needed.

**What you'll learn:**
- The 9 content rules every untrusted tool must pass (SSRF, namespaces, domains, methods, credentials, and more)
- Risk levels, and which tools ask a human on every call
- Who approves a call: `MATIMO_APPROVED_PATTERNS`, per-call and instance `on_approval`
- Execution gates by caller: draft, deprecated and production rules
- HITL quarantine by risk, the hash-chained audit log, and legacy mode

⏱️ **Time:** 15 min

**Prerequisites:** Complete Notebook 01 first

## 3️⃣ Meta-Tools - Agents That Create Tools

**[03_meta_tools.ipynb](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/03_meta_tools.ipynb)** [![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/03_meta_tools.ipynb)

Watch an AI agent **define and deploy its own tools at runtime** - the most advanced Matimo feature.

**What you'll learn:**
- Meta-tools: `matimo_create_tool`, `matimo_reload_tools`, `matimo_approve_tool`, `matimo_list_user_tools`
- The lifecycle: the agent creates a draft → you approve it → the agent uses it
- Why an agent's tool is a draft that cannot run until a human approves it
- Live demo: agent writes YAML, registers it, and uses it after your review

⏱️ **Time:** 15 min

**Prerequisites:** Complete Notebook 01 first

## 4️⃣ Skills - Progressive Knowledge for Agents

**[04_skills.ipynb](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/04_skills.ipynb)** [![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/04_skills.ipynb)

Give agents domain-specific knowledge on demand, following the [Agent Skills specification](https://agentskills.io/specification) - without stuffing every guideline into the system prompt. Zero API keys needed.

**What you'll learn:**
- The 3 progressive-disclosure levels: metadata → full content → bundled resources
- Creating a skill at runtime with `matimo_create_skill`
- Semantic discovery with `matimo_search_skills` and token-budgeted loading with `matimo_get_skill_content`
- Building a ready-to-inject system-prompt snippet with `build_relevant_skill_prompt`
- Registering skills from your own storage and loading one section at a time

⏱️ **Time:** 10 min

**Prerequisites:** Complete Notebook 01 first

## 5️⃣ MCP Server - Model Context Protocol

**[05_mcp_server.ipynb](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/05_mcp_server.ipynb)** [![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/tallclub/matimo/blob/main/docs/notebooks/05_mcp_server.ipynb)

Expose Matimo tools over the [Model Context Protocol](https://modelcontextprotocol.io/) - the same way Claude Desktop, Cursor, and Windsurf discover and call tools. Zero API keys needed.

**What you'll learn:**
- Stdio transport (`MCPServer(transport="stdio")`) - the Claude Desktop / Cursor integration path
- Connecting with the official `mcp` Python client, listing tools, and calling one
- Skills auto-exposed as MCP resources (`skills://<name>`)
- Risky tools asking the person behind the MCP client (elicitation)
- HTTP transport for remote/Docker deployments - bearer-token auth and `/health`
- Wiring a Matimo MCP server into Claude Desktop's config

⏱️ **Time:** 15 min

**Prerequisites:** Complete Notebook 01 first
